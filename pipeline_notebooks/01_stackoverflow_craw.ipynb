```python

from google.cloud import bigquery
import pandas as pd
import os
import time

os.environ['GOOGLE_APPLICATION_CREDENTIALS'] = '../../../top-chain-456408-t7-97d2306c15a0.json'
client = bigquery.Client()

# Tên file lưu trữ
OUTPUT_FILE = 'stackoverflow_full_data.csv'

# Xóa file cũ nếu tồn tại để viết mới
if os.path.exists(OUTPUT_FILE):
    os.remove(OUTPUT_FILE)

# 2. Hàm lấy data theo năm
def fetch_data_by_year(year):
    print(f"🔄 Đang tải dữ liệu năm {year}...")
    
    query = f"""
        SELECT 
            id, 
            creation_date, 
            title, 
            tags, 
            view_count, 
            score, 
            answer_count,
            comment_count
        FROM 
            `bigquery-public-data.stackoverflow.posts_questions`
        WHERE 
            EXTRACT(YEAR FROM creation_date) = {year}
            AND score >= 1  -- Chỉ lấy câu có ít nhất 1 vote (bỏ rác)
            AND tags IS NOT NULL
    """
    
    # Chạy query
    df = client.query(query).to_dataframe()
    return df

# 3. Vòng lặp chính: Chạy từ 2014 đến 2025
# (Bạn có thể sửa range nếu muốn lấy xa hơn, ví dụ 2008)
start_year = 2014
end_year = 2026

total_rows = 0

for year in range(start_year, end_year + 1):
    try:
        # Lấy dữ liệu
        df_year = fetch_data_by_year(year)
        row_count = len(df_year)
        
        if row_count > 0:
            # Xử lý datetime
            df_year['creation_date'] = pd.to_datetime(df_year['creation_date'])
            
            # Ghi vào file CSV (Mode 'a' = Append - Ghi nối tiếp)
            # Nếu là lần đầu tiên (header=True), các lần sau không ghi header nữa
            write_header = (total_rows == 0)
            df_year.to_csv(OUTPUT_FILE, mode='a', index=False, header=write_header)
            
            total_rows += row_count
            print(f"✅ Đã lưu xong năm {year}: {row_count} dòng. (Tổng: {total_rows})")
        else:
            print(f"⚠️ Năm {year} không có dữ liệu phù hợp.")
            
        # Nghỉ 1 chút để không spam request quá gắt (optional)
        time.sleep(1)
        
        # Giải phóng RAM thủ công cho chắc ăn
        del df_year

    except Exception as e:
        print(f"❌ Lỗi ở năm {year}: {e}")

print("="*30)
print(f"🎉 HOÀN TẤT! Tổng cộng đã lấy: {total_rows} dòng.")
print(f"Dữ liệu nằm tại: {OUTPUT_FILE}")
```

```python
import pandas as pd
import glob
import os

MAIN_FILE = 'stackoverflow_full_data.csv'

# Lấy danh sách tất cả file CSV trong thư mục hiện tại có dạng "202*_Q*.csv"
csv_files = glob.glob("202*_Q*.csv")

print(f"📂 Tìm thấy {len(csv_files)} file dữ liệu quý: {csv_files}")

count_success = 0

for file in csv_files:
    try:
        print(f"🔄 Đang xử lý {file}...")
        df = pd.read_csv(file)
        
        if 'creation_date' in df.columns:
            df['creation_date'] = pd.to_datetime(df['creation_date'])
        
        # header=False vì file chính đã có header rồi
        df.to_csv(MAIN_FILE, mode='a', index=False, header=False)
        
        count_success += len(df)
        print(f"   ✅ Đã thêm {len(df)} dòng.")
        
        
    except Exception as e:
        print(f"   ❌ Lỗi file {file}: {e}")

print("="*30)
print(f"🎉 TỔNG KẾT: Đã gộp thêm {count_success} dòng mới vào {MAIN_FILE}")
```
